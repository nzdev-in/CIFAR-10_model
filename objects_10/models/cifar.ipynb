```python
import torch
from torch import nn
import urllib
from PIL import Image
import numpy as np
```

```python
#Raw Model
class CifarObjectClassification(nn.Module):
    def __init__(self,**kwargs):
        super(CifarObjectClassification,self).__init__(**kwargs)
        self.conv1 = nn.Conv2d(3,32,kernel_size=3,padding=1,stride=1)
        self.start = nn.Sequential(
            nn.BatchNorm2d(32),
            nn.GELU(),
            nn.Conv2d(32,64,kernel_size=3,padding=1,stride=1),
            nn.MaxPool2d(kernel_size=4)
        )
        self.end = nn.Sequential(
            nn.BatchNorm2d(64),
            nn.GELU(),
            nn.Conv2d(64,128,kernel_size=4,padding=1,stride=1)
        )
        self.out = nn.Linear(6272,10)

    def forward(self,train_images):
        #Shape = b x c x h x w
        train_images = train_images.to(torch.float32)
        x = self.conv1(train_images)
        x = self.start(x)
        x = self.end(x)
        x = x.flatten(1)
        x = self.out(x)
        return x
```

```python
#Forcing pretrained weights

model = CifarObjectClassification()
model.load_state_dict(torch.load("cifar-10-model.pth",map_location=torch.device("cpu")))
```
